```python

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant and help me teach prompt engineering to my students."),
    ("user", "Question: {question}")
])

output_parser = StrOutputParser()
```

```python
llm =ChatGroq(
    model="openai/gpt-oss-120b"
)
```

```python
chain = prompt | llm | output_parser
```

```python

result = chain.invoke({"question": """ A store has 48 apples. They sell 1/3 in the morning and 1/4 of what's left
in the afternoon. How many apples remain?"""})
```
