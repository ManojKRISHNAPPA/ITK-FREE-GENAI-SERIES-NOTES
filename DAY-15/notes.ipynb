```python
sentences = [
    "The cat sat on the mat",
    "A feline rested on the rug",
    "The dog played in the yard",
    "I love programming in Python",
    "Python is my favorite programming language"
]
```

```python
import numpy as np

def cosine_similairty(vect1,vect2):
    dot_product = np.dot(vect1,vect2) ## cosine = a.b/|a||b|
    mag_a = np.linalg.norm(vect1)
    mag_b = np.linalg.norm(vect2)

    return dot_product/(mag_a*mag_b)
```

```python
from langchain_openai import OpenAIEmbeddings

embeddings = OpenAIEmbeddings(
    model = "text-embedding-3-small"
)

sentences = [
    "The cat sat on the mat",
    "A feline rested on the rug",
    "The dog played in the yard",
    "I love programming in Python",
    "Python is my favorite programming language"
]

embed_doc = embeddings.embed_documents(sentences)
embed_doc
```

```python
for i in range(len(embed_doc)):
    for j in range(i+1,len(embed_doc)):
        sim=cosine_similairty(embed_doc[i],embed_doc[j])
        print(f"simiality between {sentences[i]} and {sentences[j]} is {sim}")
```

```python
senetence = [
    "the dog eats cat",
    "the cat eats dog"
    "the cat sat on mat",
    "the cat sat on mat",
    "i love india",
    "i love programming in python",
    "python is my favorite programing language"
]

from langchain_openai import OpenAIEmbeddings

embeddings = OpenAIEmbeddings(
    model = "text-embedding-3-small"
)

embed_doc = embeddings.embed_documents(senetence)

for i in range(len(embed_doc)):
    for j in range(i+1,len(embed_doc)):
        sim=cosine_similairty(embed_doc[i],embed_doc[j])
        print(f"simiality between {senetence[i]} and {senetence[j]} is {sim}")
```
