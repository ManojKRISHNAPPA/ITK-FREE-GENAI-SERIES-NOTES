```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
word_embeddings = {
    "toxic": [0.8,0.6],
    "kgf": [0.85,0.65],
    "drama": [0.7,0.3],
    "kirathakaru": [0.65,0.35],
    "googly": [-0.5,0.3],
    "Rajahuli": [-0.45,0.15],
}
```

```python
fig,ax = plt.subplots(figsize=(8,6))

for word, coords in word_embeddings.items():
    ax.scatter(coords[0], coords[1], s=100)
    ax.annotate(word, (coords[0], coords[1]), xytext=(5, 5), 
                textcoords='offset points')
ax.set_xlabel('Dimension 1')
ax.set_ylabel('Dimension 2')
ax.set_title('Simplified Word Embeddings in 2D Space')
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
# Similiarity measuement:
```

```python
def cosine_similarity(vector1,vector2):

    dot_product = np.dot(vector1,vector2)
    norma_a = np.linalg.norm(vector1)
    norma_b = np.linalg.norm(vector2)

    return dot_product/(norma_a*norma_b)
```

```python
toxic=[0.8,0.6]
kgf= [0.85,0.65]
```

```python
toxic_vs_kgf_simil=cosine_similarity(toxic,kgf)
```

```python
print(toxic_vs_kgf_simil)
```

```python
toxic=[0.8,0.6]
Rajahuli= [-0.45,0.15]
```

```python
toxic_vs_kgf_simil=cosine_similarity(toxic,Rajahuli)
print(toxic_vs_kgf_simil)
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model = 'sentence-transformers/all-MiniLM-L6-v2'
    )
```

```python
text = "Hello ITK Fam how are you guys...."

embeding = embeddings.embed_query(text)

print(f"Text: {text}")
print(f"Embedding length: {len(embeding)}")
print(embeding)
```
