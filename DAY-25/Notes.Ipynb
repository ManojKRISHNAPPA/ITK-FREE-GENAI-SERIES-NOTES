## 1. Core Definitions

### AI Agents
> "AI agents" refers to individual software programs designed to perform **specific tasks** with a degree of autonomy — **without human intervention**.

- Single entity with a **focused purpose**
- Operates on **predefined rules**
- **Limited autonomy** — knows what to do, but not why or when to adapt
- Decision-making is deterministic: input → rule → output

### Agentic AI
> Agentic AI describes a **broader framework** where multiple AI agents can collaborate and make decisions independently to achieve a larger goal, emphasizing **adaptability and learning from experience** rather than just following predefined rules.

- A **network of agents** working together to achieve a goal
- Solve a task **without human intervention**
- Focuses on the **overall system** with the ability to **learn, adapt, and make complex decisions**
- **Autonomous decision-making** based on real-time data and circumstances

---

## 2. Head-to-Head Comparison

| Dimension | AI Agent | Agentic AI |
|---|---|---|
| Scope | Single entity | Multi-agent network |
| Autonomy | Limited, predefined rules | High, adaptive, context-driven |
| Decision Making | Predefined rule-based | Autonomous, real-time data |
| Goal | Specific narrow task | Broad, complex objective |
| Learning | Static (no learning) | Improves and adapts over time |
| Human Oversight | Minimal for its defined task | Human-in-the-loop at a higher level |
| Complexity | Low | High |
| Example | Customer Service Chatbot | Personalized Health Assistant |

## 3. The Four Pillars of Agentic AI

```
Perception  →  Reasoning  →  Action  →  Learning
```

| Pillar | Description |
|---|---|
| **Perception** | Gather data from surrounding environment (sensors, APIs, DBs) |
| **Reasoning** | Understand what is going on — interpret context, infer intent |
| **Action** | Take a specific action — call a tool, write code, send a message |
| **Learning** | Improve and adapt over time based on outcomes and feedback |

This PRAL loop is what distinguishes Agentic AI from simple AI Agents — agents only do **Action**, agentic systems do all four continuously.

## 4. The Brain of Every AI Agent

```
LLM Models  ==>  Brain of the AI Agent
                    │
                    ├── Tools        (APIs, web search, calculators)
                    └── Databases    (vector stores, SQL, documents)
```

The LLM is the **reasoning engine**. Tools and databases are its **hands and memory**. Every AI agent — whether standalone or part of an agentic network — has this structure at its core.

## 6. Agentic AI System — Software Project Example (XYZ Company)

This is a real-world Agentic AI applied to **software engineering**.

### Traditional Process
```
1. Requirement Gathering  →  Business Analyst, Product Manager
2. Sprint Planning        →  [Agile Process]
3. Developer Teams        →  Code, Test, Review
```

### Agentic AI System Overlay
```
                Human Intervention
                        │
                        ▼
              Requirements  ←──  LLM (Accuracy check)
                        │
            ┌───────────┼───────────┐
          Task 1      Task 2      Task 3
            │           │           │
            ▼           ▼           ▼
          Dev 1        Dev 2       Dev 3
        [AI Agent]  [AI Agent]  [AI Agent]         ──► Code Review
        {LLM}                                          [AI Agent]
        {Prompt Eng}
            │           │           │
            └───────────┴───────────┘
                        │
                        ▼
                    Testing
                   [AI Agent]
                        │
                        ▼
                     Bug Fix
                        │
                        ▼
               Human Feedback in the Loop
                        │
                        ▼
                 Coding the Project ✅
```

**What each agent does:**

| Agent | Role | Tools Used |
|---|---|---|
| Requirements Agent | Parses specs, checks ambiguity, routes tasks | LLM + doc parser |
| Dev Agent 1/2/3 | Writes code for assigned tasks | LLM + code interpreter |
| Code Review Agent | Reviews PRs, checks style, logic, security | LLM + linting tools |
| Testing Agent | Writes and runs test cases | LLM + test runner |
| Bug Fix Agent | Triages failures, suggests fixes | LLM + error logs |

**Key design patterns here:**
- **Parallelism** — Dev 1, 2, 3 run concurrently (tasks are independent)
- **Human-in-the-loop** — Humans review before final merge
- **Feedback loop** — Test failures flow back to Dev agents
- **Orchestrator** — LLM at the top routes tasks intelligently

```
Topic / Keyword Input

        │
        ▼

        │

  (drafts each section with LLM)
        │
        ▼
  SEO Agent
  (keyword density, meta description, slug)
        │
        ▼
  Editor Agent
  (grammar, tone, readability scoring)
        │
        ▼
  Human Review (optional gate)
        │
        ▼
  Published Blog Post ✅
```

Each step is a **specialized AI agent**. The pipeline is orchestrated by a controller that passes context between agents. This is Agentic AI because:
- No single agent knows how to do everything
- Output of one agent feeds the next
- System can retry, revise, and self-correct at each stage

## 9. Architecture Patterns

### Pattern 1: Sequential (Pipeline)
```
Agent A  →  Agent B  →  Agent C  →  Output
```
Used in: Blog generation, document processing, data ETL

### Pattern 2: Parallel (Fan-out / Fan-in)
```
              ┌── Agent A ──┐
Orchestrator ─┼── Agent B ──┼─► Aggregator → Output
              └── Agent C ──┘
```
Used in: Software dev (multiple devs), multi-source research

### Pattern 3: Hierarchical (Manager-Worker)
```
          Manager Agent (LLM)
         /        |         \
   Worker 1   Worker 2   Worker 3
```
Used in: Complex projects, automated pipelines with sub-tasks

### Pattern 4: Feedback Loop (Reflexion)
```
Agent → Output → Evaluator → [pass] → Done
                           → [fail] → Agent (retry with feedback)
```
Used in: Code generation, writing, reasoning tasks

## 10. LLM as the Agent Brain — Technical Detail

```python
# Minimal AI Agent structure
class AIAgent:
    def __init__(self, llm, tools, memory):
        self.llm = llm          # GPT-4.1, Gemini, Claude, etc.
        self.tools = tools      # search, code_exec, db_query, API calls
        self.memory = memory    # vector store, conversation history

    def run(self, task: str) -> str:
        context = self.memory.retrieve(task)
        plan = self.llm.plan(task, context)        # Reasoning
        for step in plan:
            result = self.tools.execute(step)      # Action
            self.memory.store(step, result)        # Learning / State
        return self.llm.synthesize(plan, results)  # Output
```

**Agentic AI adds orchestration on top:**
```python
class AgenticSystem:
    def __init__(self, agents: list[AIAgent], orchestrator_llm):
        self.agents = agents
        self.orchestrator = orchestrator_llm

    def run(self, goal: str):
        # Orchestrator breaks goal into tasks
        task_plan = self.orchestrator.decompose(goal)
        results = {}
        for task in task_plan:
            agent = self.route(task)          # pick the right agent
            results[task] = agent.run(task)   # execute
        return self.orchestrator.synthesize(results)
```

````python
import json
from typing import Any, Dict
import requests

WMO_WEATHER_CODES = {
    0: "Clear sky",
    1: "Mainly clear",
    2: "Partly cloudy",
    3: "Overcast",
    45: "Fog",
    48: "Depositing rime fog",
    51: "Light drizzle",
    53: "Moderate drizzle",
    55: "Dense drizzle",
    61: "Slight rain",
    63: "Moderate rain",
    65: "Heavy rain",
    71: "Slight snow fall",
    73: "Moderate snow fall",
    75: "Heavy snow fall",
    80: "Slight rain showers",
    81: "Moderate rain showers",
    82: "Violent rain showers",
    95: "Thunderstorm",
    96: "Thunderstorm with slight hail",
    99: "Thunderstorm with heavy hail",
}


def get_current_weather(location: str) -> str:
    """Fetch the current weather conditions for a specified city or location.

    Args:
        location: The name of the city, town, or region (e.g., 'Bengaluru',
          'Tokyo', 'New York').

    Returns:
        A JSON string containing location details, temperature (°C),
        apparent temperature (°C), humidity (%), wind speed (km/h),
        and weather condition description.
    """
    if not location or not location.strip():
        return json.dumps(
            {"error": "Please provide a valid location name."}, indent=2
        )

    session = requests.Session()

    try:
        # Step 1: Geocode the location name to coordinates
        geo_url = "https://geocoding-api.open-meteo.com/v1/search"
        geo_params = {"name": location.strip(), "count": 1, "language": "en"}
        geo_resp = session.get(geo_url, params=geo_params, timeout=5)
        geo_resp.raise_for_status()
        geo_data = geo_resp.json()

        results = geo_data.get("results")
        if not results:
            return json.dumps(
                {
                    "error": f"Location '{location}' not found. Please verify the city name."
                },
                indent=2,
            )

        top_result = results[0]
        latitude = top_result["latitude"]
        longitude = top_result["longitude"]
        resolved_name = (
            f"{top_result.get('name')}, {top_result.get('country', '')}"
        )

        # Step 2: Fetch current meteorological conditions
        weather_url = "https://api.open-meteo.com/v1/forecast"
        weather_params = {
            "latitude": latitude,
            "longitude": longitude,
            "current": [
                "temperature_2m",
                "relative_humidity_2m",
                "apparent_temperature",
                "weather_code",
                "wind_speed_10m",
            ],
            "timezone": "auto",
        }
        weather_resp = session.get(
            weather_url, params=weather_params, timeout=5
        )
        weather_resp.raise_for_status()
        weather_data = weather_resp.json()

        current = weather_data.get("current", {})
        code = current.get("weather_code")
        condition = WMO_WEATHER_CODES.get(code, "Unknown / Variable")

        output: Dict[str, Any] = {
            "query": location,
            "resolved_location": resolved_name,
            "latitude": latitude,
            "longitude": longitude,
            "temperature_celsius": current.get("temperature_2m"),
            "apparent_temperature_celsius": current.get("apparent_temperature"),
            "relative_humidity_percent": current.get("relative_humidity_2m"),
            "wind_speed_kmh": current.get("wind_speed_10m"),
            "condition": condition,
            "time": current.get("time"),
        }
        return json.dumps(output, indent=2)

    except requests.exceptions.Timeout:
        return json.dumps(
            {"error": "The weather service request timed out."}, indent=2
        )
    except requests.exceptions.RequestException as e:
        return json.dumps(
            {"error": f"Network error fetching weather data: {str(e)}"},
            indent=2,
        )
    except Exception as e:
        return json.dumps(
            {"error": f"An unexpected error occurred: {str(e)}"}, indent=2
        )
````

````python
import json
from typing import Any, Dict, List
from ddgs import DDGS
from langchain_core.tools import tool


@tool
def search_web(query: str, max_results: int = 5) -> str:
    """Search the live web for current events, news, technical documentation,

    or general facts.

    Args:
        query: The search term or question to look up on the web.
        max_results: The maximum number of search results to return (default 5).

    Returns:
        A JSON string containing a list of search results, each with 'title',
        'snippet', and 'url'.
    """
    if not query or not query.strip():
        return json.dumps(
            {"error": "Query cannot be empty. Please provide a search term."},
            indent=2,
        )

    try:
        results: List[Dict[str, Any]] = []
        with DDGS() as ddgs:
            raw_results = ddgs.text(
                query.strip(), max_results=min(max_results, 10)
            )

            if not raw_results:
                return json.dumps(
                    {
                        "message": f"No results found for query: '{query}'. Try rephrasing."
                    },
                    indent=2,
                )

            for item in raw_results:
                results.append(
                    {
                        "title": item.get("title", ""),
                        "snippet": item.get("body", ""),
                        "url": item.get("href", ""),
                    }
                )

        return json.dumps(
            {"query": query, "results": results}, indent=2, ensure_ascii=False
        )

    except Exception as e:
        return json.dumps(
            {"error": f"Failed to execute web search: {str(e)}"}, indent=2
        )
````

````python
from langchain.agents import create_agent
from langchain_core.messages import HumanMessage
agent = create_agent(
    model = "gpt-4o",
    tools = [search_web],
    system_prompt= "you are a helpfull assitant"
)

agent
````
